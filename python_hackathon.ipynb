```python
from google.colab import files
uploaded = files.upload()

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("PS22_Dataset_22.csv")
print(df.head())
print(df.info())
print(df.describe())
print(df.isnull().sum())


df = df.drop_duplicates()

df = df.fillna(method='ffill')
print(df.isnull().sum())


print("Mean:", df["A"].mean())
print("Median:", df["A"].median())

plt.figure()
plt.hist(df["A"])
plt.title("Distribution of A")
plt.xlabel("A")
plt.ylabel("Frequency")
plt.show()

plt.figure()
df.groupby("B")["A"].mean().plot(kind='bar')
plt.title("Average A by B")
plt.xlabel("B")
plt.ylabel("Average A")
plt.show()
```
